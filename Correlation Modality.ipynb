```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
heights = [
    150, 152, 155, 157, 158,
    160, 160, 161, 162, 163,
    164, 165, 165, 166, 167,
    168, 170, 172, 175, 178
]

df = pd.DataFrame({'Height':heights})

sns.histplot(df['Height'], kde = True)
plt.title('Normal Distribution')
plt.show()
```

```python
print(df['Height'].mean())
print(df['Height'].median())
```

```python
salary = [
    20, 22, 24, 25, 26,
    27, 28, 29, 30, 32,
    35, 40, 45, 60, 90
]
```

```python
df = pd.DataFrame({'Salary':salary})

sns.histplot(df['Salary'],kde = True, color= 'orange')
plt.title('Right Skewed')
plt.show()
```

```python
marks = [
    30, 45, 60, 70,
    75, 78, 80, 82,
    84, 85, 86, 88,
    90, 92, 95
]
```

```python
df = pd.DataFrame({'Marks':marks})

sns.histplot(df['Marks'],kde = True, color= 'darkorange')
plt.title('Right Skewed')
plt.show()
```

```python
ages = [
    18, 18, 19, 19, 20,
    20, 20, 21, 21, 22,
    23, 24
]
```

```python
df = pd.DataFrame({'Age':ages})

sns.histplot(df['Age'],kde = True, bins=6,color=sns.color_palette("coolwarm", 6)[3])

plt.title("Bimodal Distribution")
plt.show()
```

```python
ratings = [
    1, 1, 2, 2,
    5, 5, 6, 6,
    9, 9, 10, 10
]

df = pd.DataFrame({'Ratings': ratings})

sns.histplot(df['Ratings'], kde=True,color= 'purple')

plt.title("Multimodal Distribution")
plt.show()
```

```python
data = {
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [35, 45, 55, 70, 85]
}

df = pd.DataFrame(data)

print(df)

print("\nCovariance:")
print(df.cov())

print("\nCorrelation:")
print(df.corr())
```

```python
data = {
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Likes_Python': ['Yes', 'Yes', 'No', 'Yes', 'No', 'No']
}

df = pd.DataFrame(data)

print(df)
```

```python
cross_tab = pd.crosstab(df['Gender'], df['Likes_Python'])

print(cross_tab)
```

```python
cross_tab.plot(kind= 'bar')
```
