```python
import pandas as pd

# Dataset
data = {
    "Student": ["A","B","C","D","E","F"],
    "Gender": ["Boy","Boy","Boy","Girl","Girl","Girl"],
    "Passed": ["Yes","No","Yes","Yes","No","Yes"]
}

df = pd.DataFrame(data)

df
```

```python
total = len(df)
```

```python
total
```
