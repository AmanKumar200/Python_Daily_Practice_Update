```python
import pandas as pd

data = {
    "Student": ["A","B","C","D","E","F","G","H","I","J"],
    "Cricket": ["Yes","Yes","Yes","No","No","Yes","No","Yes","No","Yes"],
    "Passed": ["Yes","Yes","No","Yes","No","Yes","No","Yes","Yes","No"]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
play_cric= len(df[df['Cricket']=='Yes'])/len(df)
```

```python
play_cric
```

```python
passed = len(df[df['Passed']=='Yes'])/len(df)
passed
```

```python
cricket_passed= len(df[(df['Cricket']=='Yes') & (df['Passed']=='Yes')])/len(df)
cricket_passed
```

```python
cricket_or_passed = len(df[(df['Cricket']=='Yes') | (df['Passed']=='Yes')])/len(df)
```

```python
cricket_or_passed
```

### probability of passed given he plays cricket

```python
passed_given_cricket = cricket_passed/play_cric
passed_given_cricket 
```

## Bases Theorem- probability of cricket given passed

```python
cricket_given_passed = passed_given_cricket * play_cric/passed
```

```python
cricket_given_passed
```
