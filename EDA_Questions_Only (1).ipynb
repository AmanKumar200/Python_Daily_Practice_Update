### Question 26
Records by Region

```python
Records_by_Region = (df.groupby('region').size().reset_index(name='record_count'))
plt.figure(figsize = (10,5))
sns.barplot(data=Records_by_Region, x='region', y='record_count', palette="Set2")
plt.title("Records by Region")
plt.xlabel("Region")
plt.ylabel("Record Count")
plt.xticks(rotation=30)
plt.show()
```

### Question 27
AI Adoption by Region

```python
ai_by_region = (df.groupby('region')['ai_adoption_index'].mean().reset_index().sort_values(by='ai_adoption_index', 
                                                                                           ascending=False))
plt.figure(figsize=(8,5))
sns.barplot(data=ai_by_region, x='ai_adoption_index', y='region', palette="viridis")

plt.title("AI Adoption by Region")
plt.xlabel("Average AI Adoption Index")
plt.ylabel("Region")
plt.show()
```

### Question 28
AI Adoption Over Time

```python
avg_ai_by_year = (
    df.groupby('year')['ai_adoption_index']
      .mean()
      .reset_index()
)
plt.figure(figsize=(8,5))
sns.lineplot(data=avg_ai_by_year, x='year', y='ai_adoption_index',
             marker='o', color="orange", linewidth=2)
for i, row in avg_ai_by_year.iterrows():
    plt.text(row['year'], row['ai_adoption_index']+0.01,
             f"{row['ai_adoption_index']:.2f}", ha='center')

plt.title("AI Adoption Over Time")

plt.title("AI Adoption Over Time")
plt.xlabel("Year")
plt.ylabel("Average AI Adoption Index")
plt.show()
```

### Question 34
Automation Risk by Industry (Bar Chart)

```python


# Average automation risk per industry
automation_risk = (
    df.groupby('industry_sector')['sector_automation_risk_score']
      .mean()
      .reset_index()
      .sort_values(by='sector_automation_risk_score', ascending=False)
)

# Bar plot
plt.figure(figsize=(10,6))
sns.barplot(data=automation_risk, x='industry_sector', y='sector_automation_risk_score', palette="Set2")

# Add exact value callouts
for i, row in automation_risk.iterrows():
    plt.text(i, row['sector_automation_risk_score']+0.01,
             f"{row['sector_automation_risk_score']:.3f}", ha='center')

plt.title("Automation Risk by Industry")
plt.xlabel("Industry Sector")
plt.ylabel("Average Automation Risk Score")
plt.xticks(rotation=30)
plt.show()
```
