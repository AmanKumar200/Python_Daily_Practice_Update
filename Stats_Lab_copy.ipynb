```python
sample_row = df.sample(n=40,random_state= 42, replace = False)
sample_mean = sample_row['salary'].mean()
print(f"Sample Mean: {sample_mean}")
```

**Exercise 3:** Perform **Stratified Sampling** by `department`: sample 10% from each stratum (at least 2 per department). Compute the combined sample mean salary.

```python
def stratified_sampling(group):
    sample_size = int(len(group)*0.1)  # Adjust the fraction as needed
    return group.sample(sample_size , random_state = 2)
stratified_sample = df.groupby('department').apply(stratified_sampling)
stratified_sample['salary'].mean()
```

# Dispersion: Range, Variance/Std (Population vs Sample), MAD, CV

**Exercise 1:** Compute the **range** of `salary`.

```python
#range = max- min
salary_range = df['salary'].max() - df['salary'].min()
print(f"Salary Range: {salary_range}")
```

**Exercise 2:** Compute **population variance** and **population std dev** of `salary`.

```python
population_variance = df['salary'].var(ddof=0)
print(f"Population Variance: {population_variance}")
population_std = df['salary'].std(ddof=0)
print(f"Population_std: {population_std}")
```

**Exercise 3:** Compute **sample variance** and **sample std dev** of `salary`.

```python
sample_variance = df['salary'].var(ddof=1)
print(f"Sample Variance: {sample_variance}")
sample_std = df['salary'].std(ddof=1)
print(f"Sample_std: {sample_std}")
```

**Exercise 4:** Compute **Mean Absolute Deviation (MAD)** around the mean for `salary`.

```python
mean_salary = df['salary'].mean()
mad_salary = abs(df['salary'] - mean_salary).mean()
MAD= np.mean(abs(df['salary'] - mean_salary))
print(f"Mean Salary: {mean_salary}")
print(f"Mean Absolute Deviation: {mad_salary}")
print(f"MAD: {MAD}")
```

**Exercise 5:** Compute **Coefficient of Variation (CV)** for `salary` and `monthly_sales` (population std / mean).

```python
salary_CV = (df['salary'].std(ddof=0)/df['salary'].mean())
print(f"salary's CV is: {salary_CV}")
monthly_sales_CV = (df['monthly_sales'].std(ddof=0)/df['monthly_sales'].mean())
print(f"monthly_sales's CV is: {monthly_sales_CV}")
```

**Exercise 6:** By **department**, compute sample std dev and CV of `salary`.

```python
print("--- Computed the standard deviation  for each department ---")
df.groupby('department')['salary'].std(ddof=1).reset_index()
```

```python
print("--- Computed the CV for each department ---")
std = df.groupby('department')['salary'].std(ddof=1)
mean = df.groupby('department')['salary'].mean()
CV_for_departments = std/mean
CV_for_departments.reset_index()
```
